# 3 Imprimir Números pares e impares

# Elabore um código para imprimir todos os números pares e ímpares de 1 a 20 usando for

```python
for numero in range (1,21):
    if numero %2 == 0:
        print (f" O {numero} é par")
    else:
        print (f" O {numero} é impar")
```

```python
print (f"Os numeros pares de 1 a 0 são:")
for numero in range(1,21):
    if (numero % 2 == 0):
        print (f"{numero:2d}", end=",")
```

```python
print(f"Os numeros impares de 1 a 20 são:")
for numero in range (1,21):
    if (numero % 2 == 1):
      print (f"{numero:2d}", end=",")
```

```python
pares=[]
impares=[]
for numero in range (1,21):
    if (numero % 2 == 0):
        pares.append(numero)
    else:
        impares.append(numero)
print(f"Numeros pares : {pares}")
print(f"Numeros impares : {pares}")
```

# Imprimir de 10 a 1 utilizando while

```python
numero=10
while numero > 0:
    print(f'{numero}')
    numero-=1
```

## 5️⃣ Criar um programa que pergunta o nome

```python
nome=""
while nome.lower() != "sair":
    nome=input("Digite um nome:")
    print (f"seu nome é {nome}")
    if nome.lower() != "sair":
        print(f"Olá{nome}")
```

#  Imprimir a soma dos números

```python
soma=0
for numero in range (1,101):
    soma= soma + numero
print (f"O valor da soma de todos o numeros de 1 até 100 é {soma}")
```

```python
soma=0
numero=1
while numero <= 100:
    soma = soma + numero
    numero = numero + 1
    print (f'{numero}')
    print (f'a soma dos numeros é {soma}')
```
