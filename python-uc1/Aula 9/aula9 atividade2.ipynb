## 1️⃣ Cadastro de produtos

```python
produtos=(
    ("Arroz", 5.99),
    ("Feijão", 7.49),
    ("Leite", 4.89),
    ("Oleo", 9.99),
    ("Açucar", 3.29)
)
print("LISTA DE PRODUTOS")
for nome, preco in produtos:
    print(f"{nome:_<20} R$ {preco:.2f}")
```

## 2️⃣ Média de notas

```python
notas=(7.5,8.0,6.5,9.0)
media=sum(notas)/len(notas)
print(f"Média das notas:{media:.2f}")
```

## 3️⃣ Verificação de números primos

```python
numeros=(2,7,10,13,17,22,29,31,40,53)
def eh_primo(n):
    if n<2:
        return False
    for i in range(2, int(n**0.5)+1):
        if n%i==0:
            return False
    return True

primos=tuple(num for num in numeros if eh_primo(num))
print(f"Números primos: {primos}")
```

## 4️⃣ Classificação de idade

```python
idades=(18,25,16,30,14)
maiores=sum(1 for idade in idades if idade>=18)
print(f"Quantidade de maiores de idade: {maiores}")
```

## 5️⃣ Palavras com vogais

```python
palavras=("python", "programação", "desenvolvimento", "computador")
for palavra in palavras:
    vogais="".join(letra for letra in palavra if letra in "aeiou")
    print(f"Vogais de {palavra}:{vogais}")
```

## 6️⃣ Números por extenso

```python
extenso=("zero","um","dois","três","quatro","cinco","seis","sete","oito","nove")
num=int(input("Digite um numero(0-9): "))
if 0 <= num <=9:
    print(f" Você digitou: {extenso[num]}")
else:
    print("Número inválido.")
```

## 7️⃣ Maior e menor valor

```python
import random
numeros=tuple(random.randint(1,100)for _ in range(10))
print(f"Números: {numeros})")
print(f"Maior: {max(numeros)}, menor:{min(numeros)}")
```

## 8️⃣ Frequência de elementos

```python
numeros=(3,7,2,3,9)
n=int(input("Digite um numero para verificar: "))
print(f"O numero{n}aparece{numeros.count(n)} vezes.")
```

## 9️⃣ Ordenação sem modificação da tupla

```python
numeros=(10,3,7,1,9)
ordenados=tuple(sorted(numeros))
print(f"Tupla original: {numeros}")
print(f"Tupla Ordenada: {ordenados}")
```

## 🔟 Jogo de dados

```python
import random

lancamentos = tuple(random.randint(1, 6) for _ in range(5))
print(f"Resultados: {lancamentos}")

for num in range(1, 7):
    print(f"{num} saiu {lancamentos.count(num)} vezes.")
```

```python
import random

idades=[]
for _ in range (100):
    idades.append(random.randint(1,92))

maiores = 0
for idade in idades:
    if idade >= 18:
        maiores += 1

print(f"Quantidade de maiores de idade: {maiores}")
```

```python
import random

# Gerando os 5 lançamentos com for tradicional
lancamentos_lista = []
for _ in range(5):
    lancamentos_lista.append(random.randint(1, 6))

lancamentos = tuple(lancamentos_lista)
print(f"Resultados: {lancamentos}")

# Contagem tradicional
for num in range(1, 7):
    print(f"{num} saiu {lancamentos.count(num)} vezes.")
```
