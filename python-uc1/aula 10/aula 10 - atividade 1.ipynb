```python
matriz = []

print("Digite os valores da matriz 4x4:")


for i in range(4):
    linha = []  
    
  
    for j in range(4):
        valor = int(input(f"Digite o valor para a posição [{i}][{j}]: "))
        linha.append(valor)  
    
    matriz.append(linha)  

print("\nMatriz 4x4:")
for linha in matriz:
    for elemento in linha:
        print(f"{elemento:4}", end=" ")
    print() 
```

```python
import random

def criar_matriz(qt_linhas=4, qt_colunas=4):
    matriz=[]
    for _LINHA in range (qt_linhas) :
        linha=[]
        for _COLUNA  in range (qt_colunas) :
            linha.append(random.randint(1,100))
        matriz.append(linha)  

    return (matriz)

matriz=criar_matriz(5,5)

for linha in matriz:
    print(linha)
```

```python
def imprime_matriz(matriz=[]):
    for linha in matriz:        
        for elemento in linha :
            print(f"{elemento:4d}",end=" ")
        print()    
```

```python
minha_matriz=criar_matriz(6,7)
```

```python
imprime_matriz(minha_matriz)
```
