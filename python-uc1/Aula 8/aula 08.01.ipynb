# Exercício 3: Função par_ou_impar

```python
def par_ou_impar(n):
    if n%2==0:
        return "Par"
    else:
        return "Ímpar"
print("O número 4 é", par_ou_impar(4))
print("O numero 7 é", par_ou_impar(7))
```

# Exercício 4: Função fatorial

```python
def fatorial(n):
    if n<0:
        return "Número invalido para fatorial."
    elif n==0:
        return 1
    else:
        resultado=1
        for i in range(1, n+1):
            resultado*=i
            return resultado
print("Fatorial de 5:", fatorial(5))
```

# Exercício 5: Função maior_elemento

```python
def maior_elemento(lista):
    return max(lista)
numeros = [3, 8, 1, 20, 7]
print("O maior elemento é:", maior_elemento(numeros))
```

# Exercício 7: Função eh_palíndromo

```python
def eh_palindromo(palavra):
    return palavra==palavra[::-1]
print("ana é palíndromo?", eh_palindromo("ana"))
print("python é palíndromo?", eh_palindromo("python"))
```

# Exercício 8: Função contar_vogais

```python
def contar_vogais(texto):
    vogais= "aeiouAEIOU"
    contador=0
    for char in texto:
        if char in vogais:
            contador +=1
    return contador

print("Numero de vogais em 'Python': ", contar_vogais("Python"))
```

# Exercício 9: Função tabuada

```python
def tabuada(n):
    resultado=[]
    for i in range(1, 11):
        resultado.append(n*i)
    return resultado
print("Tabuada do 5", tabuada(5))
```

# Exercício 10: Função ordenar_por_segundo

```python
def ordenar_por_segundo(listas_tuplas):
    return sorted(listas_tuplas, key=lambda x: x[0])
lista_exemplo=[("a", 3), ("b", 1), ("c",2)]
print("Lista ordenada:",ordenar_por_segundo(lista_exemplo))
```
